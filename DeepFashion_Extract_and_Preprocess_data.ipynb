```python
import os
import json
from collections import Counter

annos_path = "train/annos"
counter = Counter()

for file in os.listdir(annos_path):
    if file.endswith(".json"):
        with open(os.path.join(annos_path, file)) as f:
            data = json.load(f)

        for key in data:
            if key.startswith("item"):
                cat = data[key]["category_id"]
                counter[cat] += 1

print(counter.most_common(10))
```

```python
top5 = [cat for cat, _ in counter.most_common(5)]
print("Top 5 categories:", top5)
```

```python
import shutil

os.makedirs("filtered_dataset/annos", exist_ok=True)
os.makedirs("filtered_dataset/image", exist_ok=True)
```

```python
images_path = "train/image"

for file in os.listdir(annos_path):
    if file.endswith(".json"):
        with open(os.path.join(annos_path, file)) as f:
            data = json.load(f)

        filtered_items = {}

        for key in data:
            if key.startswith("item") and data[key]["category_id"] in top5:
                filtered_items[key] = data[key]

        if filtered_items:
            # Save filtered JSON
            with open(os.path.join("filtered_dataset/annos", file), "w") as f:
                json.dump(filtered_items, f)

            # Copy corresponding image
            img_name = file.replace(".json", ".jpg")
            shutil.copy(
                os.path.join(images_path, img_name),
                os.path.join("filtered_dataset/image", img_name)
            )
```
